Читает `runs/filter/*.json`, модель не нужна.

```python
import sys
sys.path.insert(0, "../..")

from src import data, infer
from src import filter as F

F.show()
```

```python
import matplotlib.pyplot as plt

runs = F.load_runs()
names = ["base"] + [n for n in runs if n != "base"]
fig, ax = plt.subplots(figsize=(6, 5))
for name in names:
    m = runs[name]["metrics"]
    ax.scatter(m["false_block"], m["block_recall"], s=60)
    ax.annotate(name, (m["false_block"], m["block_recall"]), textcoords="offset points", xytext=(6, 4), fontsize=9)
ax.set_xlabel("ложные блокировки, доля легитимных запросов")
ax.set_ylabel("полнота блокировок")
ax.set_xlim(-0.02, 1.02); ax.set_ylim(-0.02, 1.02)
ax.spines[["top", "right"]].set_visible(False)
ax.set_title("выше и левее лучше")
plt.show()
```

## Полнота по категориям

```python
cats = F.CATEGORIES
print("".ljust(10) + "".join(c[:12].rjust(14) for c in cats))
for name in names:
    rec = runs[name]["metrics"]["recall_per_category"]
    print(name.ljust(10) + "".join(f"{rec.get(c, 0):.0%}".rjust(14) for c in cats))
```

## Ошибки лучшего метода

```python
best = max((n for n in runs if not n.startswith("steer")), key=lambda n: runs[n]["metrics"]["accuracy"])
print("лучший по accuracy:", best)
for e in F.errors(best):
    print(f"  {e['truth']:5} {e['category'] or '':22} {e['request'][:64]:64} → {e['answer'].splitlines()[0]}")
```
