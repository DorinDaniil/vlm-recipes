# 02 · Базовая линия

`F.evaluate` генерирует метку для каждой из 101 тестовой ситуации, парсит первую строку, считает
метрики и perplexity с preference accuracy на dev, пишет `runs/filter/<имя>.json`. `F.show` печатает
таблицу с базой первой строкой и изменением в скобках.

| метрика | что считает | лучше |
|---|---|---|
| accuracy | доля верных PASS / BLOCK | выше |
| полнота BLOCK | доля нарушений, помеченных BLOCK | выше |
| ложный BLOCK | доля легитимных запросов, помеченных BLOCK | ниже |
| на ловушках | то же только на ловушках | ниже |
| категория | верная категория среди верно заблокированных | выше |
| формат | ответ начинается с PASS или BLOCK | выше |

На 101 ситуации интервал Уилсона около ±10 пунктов.

```python
import sys
sys.path.insert(0, "../..")

from src import data, infer
from src import filter as F

model, tokenizer = infer.load_model()
result = F.evaluate(model, tokenizer, "base", note="base model, no training")
F.show()
```
