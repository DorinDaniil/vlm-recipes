## Регрессии к базе

Где судья был PASS и стал FAIL, и где пропал обязательный отказ.

```python
for name in report.load_runs("extended"):
    if name != "base":
        lost = report.regressions("base", name)
        print(f"{name:10} {len(lost)} регрессий", ("· " + "; ".join(lost[:6])) if lost else "")
```

## По категориям

Средняя доля выполненных проверок; рост среднего иногда куплен просадкой в одной категории.

```python
runs = report.load_runs("extended")
categories = sorted({c for run in runs.values() for c in run["metrics"]["per_category"]})
print("категория".ljust(28) + "".join(n.rjust(11) for n in runs))
for c in categories:
    print(c.ljust(28) + "".join(f"{runs[n]['metrics']['per_category'].get(c, float('nan')):.0%}".rjust(11) for n in runs))
```
